### Ejercicio 1

$$x1+x2=7$$
 
$$−2x1+5x2=0$$

- $X0=(0,0)$

Se nota que el método de **Gauss-Seidel** tiene una trayectoria más rápida hacia la solución en comparación con Jacobi, debido a que utiliza los valores actualizados inmediatamente dentro de la misma iteración.

```python
# Cuando x0=(0,0)
A1 = np.array([[1, 1], 
               [-2, 5]], dtype=float)
b1 = np.array([7, 0], dtype=float)
sol_real_1 = (5, 2)
params = {'tol': 1e-6, 'max_iter': 20}

x0_a = [0, 0]
plot_trajectory_2d(
    [gauss_jacobi(A=A1, b=b1, x=x0_a, **params), 
     gauss_seidel(A=A1, b=b1, x=x0_a, **params)], 
    ['Jacobi', 'Gauss-Seidel'], 
    f'Ejercicio 1: {x0_a}',
    sol_real_1
)
```

- $X0=(5,2)$

Dado que el punto inicial $(5, 2)$ coincide exactamente con la solución del sistema, el error inicial es cero. Por lo tanto, los métodos no realizan ningún desplazamiento.

```python
# x0=(5, 2)
x0_b = [5, 2]
plot_trajectory_2d(
    [gauss_jacobi(A=A1, b=b1, x=x0_b, **params), 
     gauss_seidel(A=A1, b=b1, x=x0_b, **params)], 
    ['Jacobi', 'Gauss-Seidel'], 
    f'Ejercicio 1: {x0_b}',
    sol_real_1
)
```

## Ejercicio 3

$$
A = \begin{bmatrix}
1 & -1 & 2 & 0 \\
-1 & 1 & -1 & 3 \\
2 & -1 & 1 & -1 \\
0 & 3 & -1 & 2
\end{bmatrix}
, \quad
b = \begin{bmatrix}
1360 \\
1130 \\
1350 \\
3650
\end{bmatrix}
$$

**Prueba 3 posiciones iniciales, encuentre una en la que diverge el sistema.**

- Desde el inicio podemos ver que la matriz no es diagonal dominante, por lo que probablemente el sistema si diverge.

```python
A3 = np.array([[1, -1, 2, 0],
               [-1, 1, -1, 3],
               [2, -1, 1, -1],
               [0, 3, -1, 2]], dtype=float)

b3 = np.array([1360, 1130, 1350, 3650], dtype=float)

# evitar overflow en caso en que diverge
params_3 = {'tol': 1e-6, 'max_iter': 10}
```

Se realizan las 3 pruebas:

```python
x0_1 = [0, 0, 0, 0]
hist_1 = gauss_seidel(A=A3, b=b3, x=x0_1, **params_3)

x0_2 = [1, 1, 1, 1]
hist_2 = gauss_seidel(A=A3, b=b3, x=x0_2, **params_3)

x0_3 = [100, -50, 20, 10]
hist_3 = gauss_seidel(A=A3, b=b3, x=x0_3, **params_3)
```

```python
plot_trajectory_2d(
    [hist_1, hist_2, hist_3], 
    ['Prueba 1', 'Preba 2', 'Prueba 3'], 
    'Ejercicio 3',
    real_solution=None # No conocemos la solución y probablemente no converge
)

print(f"Valor final Prueba 1 (iter {len(hist_1)}):\n {hist_1[-1]}")
```
